# Configurations

```python
# Run configuration
SRC_LANG = 'en' # 'en' | '<any_lang>'
TGT_LANG = 'tr' # 'en' | '<any_lang>'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# Model configuration
MODEL_ID = 'FacebookAI/xlm-roberta-base'
# ---
LEGAMEX_TASK_ID = 'alxxtexxr/XLM-R-Base-squad-en-15K-s42-LegameX-LoRA-v261009193519'
LEGAMEX_TASK_CKPT_STEP = 7800
LEGAMEX_LANG_ID = 'alxxtexxr/XLM-R-Base-wikipedia-tr-3.5K-s42-LegameX-LoRA-v261010161203'
LEGAMEX_LANG_CKPT_STEP = 400

# Merging configuration.
# The merged weight is W = W_base + SUM_i c_i * d_i, where d_i is each model's RAW
# transfer delta and c_i is a per-model coefficient derived from its gate g_i:
#   'add': c_i = g_i        sum of gated deltas       (SUM g_i*d_i)
#   'avg': c_i = a_i * g_i  scalar-weighted avg       (SUM a_i*g_i*d_i, a=[0.5,0.5])
#   'gna': c_i = g_i / SUM  Gate-Normalized Weighted  (SUM (g_i/SUMg)*d_i, SUM_i c_i=1)
#   'nga': c_i = 1.0        no-gate addition        (SUM d_i; gates omitted)
#   'shrink': c_i = 0.5 + SHRINK_LAMBDA*(g_i - 0.5)  pull gates toward the middle (0->all 0.5, 1->as-is)
#   'moat': task gates only: c = g + beta*(g-0.5)*exp(-(g-0.5)^2/(2*sigma^2))  carve out the 0.5 spike, keep the tails & curve
#   refavg (this notebook): the SAME c_i per-model deltas PLUS the shared
#     reference-LoRA delta merged once, at the average of the two models' ref
#     gates: W = W_base + SUM_i c_i*d_i + (mean_i (1 - c_i)) * d_ref_avg
ADDITION_TYPE = 'add' # 'add' | 'avg' | 'gna' | 'nga' | 'shrink' | 'moat'
SHRINK_LAMBDA = 0.5  # 'shrink' only: pull gates toward 0.5 by this fraction.
#   1.0 -> gates exactly as learned   0.5 -> tails halfway to 0.5   0.0 -> every gate exactly 0.5
#   Sweep: 1.0 / 0.75 / 0.5 / 0.25 / 0.0
MOAT_BETA = 2.0     # 'moat' only: center-carving strength (keep <= 2.2 so the map stays monotone)
MOAT_SIGMA = 0.04   # 'moat' only: half-width of the window carved around 0.5 (larger = deeper canyon)
#   c = g + MOAT_BETA*(g-0.5)*exp(-(g-0.5)^2/(2*MOAT_SIGMA^2)), applied to the TASK gate ONLY.
#   Carves OUT the artificial 0.5 spike (mass stuck at sigmoid(0)) while leaving the left/right
#   tails and the smooth curve between them essentially untouched (identity far from 0.5).
#   Sweep: sigma {0.02, 0.04, 0.08} x beta {1.0, 2.0}

assert ADDITION_TYPE in ['add', 'avg', 'gna', 'nga', 'shrink', 'moat'], \
    f"Invalid addition type: {ADDITION_TYPE}. Must be 'add', 'avg', 'gna', 'nga', 'shrink', or 'moat'."
if ADDITION_TYPE == 'add':
    addition_weights = [1.0, 1.0]
elif ADDITION_TYPE == 'avg':
    addition_weights = [0.5, 0.5]
elif ADDITION_TYPE == 'nga':
    # 'nga' (no-gate addition): the learned gates are OMITTED at merge. Each
    # model's RAW transfer delta is added at full strength (c_i = 1.0 for every
    # weight, i.e. both gates set to 100%). Scalar weights are unused here.
    addition_weights = None
elif ADDITION_TYPE == 'shrink':
    # 'shrink': no scalar weights; the coefficients are the gates pulled toward
    # 0.5 by SHRINK_LAMBDA (handled inside compute_merge_coefficients).
    addition_weights = None
elif ADDITION_TYPE == 'moat':
    # 'moat': no scalar weights; the TASK gate is center-carved (0.5 spike removed)
    # with MOAT_BETA/MOAT_SIGMA inside compute_merge_coefficients; lang gate is
    # returned unchanged.
    addition_weights = None
else:  # 'gna': coefficients are gate-normalized; scalar weights unused
    addition_weights = None

# Determine the hub merged model ID
assert SRC_LANG in LEGAMEX_TASK_ID and TGT_LANG in LEGAMEX_LANG_ID, \
    "LegameX IDs do not match the specified source language and target language."
model_id_0, model_id_1 = LEGAMEX_TASK_ID.split(SRC_LANG)
model_id_1_0 = model_id_1.split("K-")[-1].split("-v")[0]
if ADDITION_TYPE == 'shrink':
    type_tag = f"{ADDITION_TYPE}-lam{SHRINK_LAMBDA:g}"
elif ADDITION_TYPE == 'moat':
    type_tag = f"{ADDITION_TYPE}-b{MOAT_BETA:g}-s{MOAT_SIGMA:g}"
else:
    type_tag = ADDITION_TYPE
hub_merged_model_id = f"{model_id_0}{TGT_LANG}-{model_id_1_0}-{type_tag}-refavg-v{datetime.now().strftime('%y%m%d%H%M%S')}"
print(f"Hub merged model ID: {hub_merged_model_id}")
```

# Utilities

```python
# LoRA utilities
def download_hf_model(
        repo_id: str, 
        ckpt_step: int, 
        max_checkpoints: int = 10_000,
        ckpt_interval: int = 25,
    ) -> tuple[Path, Path]:
    assert len(repo_id.split('/')) == 2, \
        f"Invalid repo_id: {repo_id}. Must be in the format 'username/repo_name'."
    local_dir = Path(repo_id.split('/')[-1])
    ignore_checkpoints = None
    
    if ckpt_step is not None:
        ignore_checkpoints = [
            f'checkpoint-{i}/*' for i in range(0, max_checkpoints, ckpt_interval) 
            if i != ckpt_step
        ]
        ignore_checkpoints += ['runs/*', '**/wandb/*']

    snapshot_download(
        repo_id=repo_id,
        local_dir=local_dir,
        ignore_patterns=ignore_checkpoints,
    )

    ckpt_dir = None
    if ckpt_step is not None:
        ckpt_dir = local_dir / f'checkpoint-{ckpt_step}'
    return local_dir, ckpt_dir

# Gate/LoRA schedule utilities.
# These mirror the schedule math in notebooks/legamex_training_xlmr.ipynb
# (LegamexLora._scheduler_progress / gate_scheduler / gate_offset_scheduler /
# _init_lora_module scaling) so that a merge can reconstruct the EXACT gate
# that was used during training from the checkpoint's legamex_config.json.
def _scheduler_progress(step: int, num_steps: int, scheduler_type: str = 'cosine') -> float:
    """Interpolation factor in [0, 1] following scheduler_type over num_steps.
    Mirrors LegamexLora._scheduler_progress."""
    if num_steps <= 0:
        return 1.0
    
    t = min(max(step / num_steps, 0.0), 1.0)
    
    if scheduler_type == 'linear':
        p = t
    elif scheduler_type == 'cosine':
        p = 0.5 * (1.0 - math.cos(math.pi * t))
    elif scheduler_type == 'quadratic':
        p = t ** 2
    elif scheduler_type == 'cubic':
        p = t ** 3
    elif scheduler_type == 'sqrt':
        p = math.sqrt(t)
    else:
        raise ValueError(
            f"Unknown gate scheduler type: {scheduler_type}. "
            "Must be one of ['linear', 'cosine', 'quadratic', 'cubic', 'sqrt']."
        )
    
    return p

def gate_scheduler_alpha(legamex_config: dict, training_step: int) -> float:
    """Gate warmup factor (0 -> 1 over gate_warmup_steps).
    Mirrors LegamexLora.gate_scheduler()."""
    warmup_steps = legamex_config.get('gate_warmup_steps', 0)
    scheduler_type = legamex_config.get('gate_scheduler_type', 'cosine')
    return _scheduler_progress(training_step, warmup_steps, scheduler_type)

def gate_offset_scheduled(legamex_config: dict, training_step: int) -> float:
    """Scheduled gate-offset term: gate_offset_init annealed to 0.0 over
    gate_offset_steps (constant gate_offset_init when steps <= 0).
    Mirrors LegamexLora.gate_offset_scheduler().
    The LEARNED term (gate.offset in legamex.pt) is separate and added on top,
    i.e. gate_offset_total = gate_offset_scheduled + gate_offset_learned."""
    init = float(legamex_config.get('gate_offset_init', -4.0))
    steps = int(legamex_config.get('gate_offset_steps', 0))
    if steps <= 0:
        return init
    p = _scheduler_progress(training_step, steps, legamex_config.get('gate_scheduler_type', 'cosine'))
    return init + (0.0 - init) * p

def ref_lora_scaling(legamex_config: dict) -> float:
    """Reference-LoRA scaling alpha / r (alpha / sqrt(r) when use_rslora).
    Mirrors LegamexLora._init_lora_module's scaling."""
    ref_lora_config = legamex_config['ref_lora_config']
    r = ref_lora_config['r']
    alpha = ref_lora_config['lora_alpha']
    return alpha / math.sqrt(r) if ref_lora_config.get('use_rslora', False) else alpha / r

def tfr_lora_scaling(legamex_config: dict) -> float:
    """Transfer-LoRA scaling alpha / r (alpha / sqrt(r) when use_rslora).
    Mirrors LegamexLora._init_lora_module's scaling."""
    tfr_lora_config = legamex_config['tfr_lora_config']
    r = tfr_lora_config['r']
    alpha = tfr_lora_config['lora_alpha']
    return alpha / math.sqrt(r) if tfr_lora_config.get('use_rslora', False) else alpha / r
```

```python
_, legamex_lang_dir = download_hf_model(repo_id=LEGAMEX_LANG_ID, ckpt_step=LEGAMEX_LANG_CKPT_STEP)
legamex_lang_file = legamex_lang_dir / 'legamex.pt'
legamex_lang_state_dict = torch.load(legamex_lang_file, map_location='cpu')
# legamex_config.json holds the merge-time constants (gate_offset_init/steps,
# gate_warmup_steps, tfr lora_alpha/r) that are NOT stored in legamex.pt
with open(legamex_lang_dir / 'legamex_config.json', 'r') as f:
    legamex_lang_config = json.load(f)

_, legamex_task_dir = download_hf_model(repo_id=LEGAMEX_TASK_ID, ckpt_step=LEGAMEX_TASK_CKPT_STEP)
legamex_task_file = legamex_task_dir / 'legamex.pt'
legamex_task_state_dict = torch.load(legamex_task_file, map_location='cpu')
with open(legamex_task_dir / 'legamex_config.json', 'r') as f:
    legamex_task_config = json.load(f)
```

```python
def fix_legamex_state_dict(
    state_dict: dict[str, torch.Tensor],
    legamex_config: dict,
    training_step: int,
) -> tuple[OrderedDict[str, torch.Tensor], OrderedDict[str, torch.Tensor], OrderedDict[str, torch.Tensor]]:
    """Reconstruct each LegameX module's RAW transfer delta, its reference-LoRA
    delta, and its gate.

    For every wrapped Linear module this recovers, from the checkpoint, the three
    quantities that compose a fully-trained LegameX module:

        d_tfr = tfr_scaling * (B_tfr @ A_tfr)     (raw transfer delta, ungated)
        d_ref = ref_scaling * (B_ref @ A_ref)     (raw reference-LoRA delta)
        g_i   = gate_scheduler() * sigmoid(B_g @ A_g + offset_sched + offset_learned)

    mirroring LegamexLora's forward / merge_and_unload in
    legamex_training_xlmr.ipynb, where each module's effective weight is

        W_eff = W_base + (1 - g_i) * d_ref + g_i * d_tfr

    The ref branch is the frozen shared reference (loaded from the pretrained
    reference LoRA at init), so it is reconstructed and merged alongside the
    transfer delta.

    Returns:
        delta_state_dict: per-module raw transfer delta d_tfr, plus qa_outputs
                          passthrough (FULL trained head weights, not deltas).
        ref_state_dict:   per-module raw reference-LoRA delta d_ref.
        gate_state_dict:  per-module gate g_i.
    """
    ref_scaling = ref_lora_scaling(legamex_config)
    tfr_scaling = tfr_lora_scaling(legamex_config)
    offset_scheduled = gate_offset_scheduled(legamex_config, training_step)
    gate_alpha = gate_scheduler_alpha(legamex_config, training_step)
    print(
        f"tfr_scaling={tfr_scaling}, "
        f"gate_offset_scheduled={offset_scheduled}, "
        f"gate_scheduler_alpha={gate_alpha} "
        f"(training_step={training_step})"
    )
    
    # The scheduled offset shapes the sigmoid, hence the gate RATIOS that GNA
    # depends on, so it must be reconstructed. The warmup alpha is a common
    # factor across all gates and cancels in the normalization w_i = g_i / Σg,
    # but we still warn if the offset has not annealed (it does NOT cancel).
    if int(legamex_config.get('gate_offset_steps', 0)) > 0 and offset_scheduled != 0.0:
        print(
            "Warning: the scheduled gate offset has not annealed to 0 at this "
            "checkpoint; the reconstructed gates (and thus the GNA weights) will "
            "differ from the fully-trained ones."
        )
    
    ref_state_dict = OrderedDict()
    delta_state_dict = OrderedDict()
    gate_state_dict = OrderedDict()

    lora_tfr_A_weight = None
    lora_tfr_B_weight = None
    lora_ref_A_weight = None
    lora_ref_B_weight = None
    gate_A_weight = None
    gate_B_weight = None
    gate_offset = None

    for key, value in state_dict.items():
        key_fixed = key.replace('__DOT__', '.')
        
        if 'lora.tfr.A' in key_fixed:
            lora_tfr_A_weight = value
        elif 'lora.tfr.B' in key_fixed:
            lora_tfr_B_weight = value
        elif 'lora.ref.A' in key_fixed:
            lora_ref_A_weight = value
        elif 'lora.ref.B' in key_fixed:
            lora_ref_B_weight = value
        elif 'gate.A' in key_fixed:
            gate_A_weight = value
        elif 'gate.B' in key_fixed:
            gate_B_weight = value
        elif 'gate.offset' in key_fixed:
            gate_offset = value
        elif 'qa_outputs' in key_fixed:
            delta_state_dict[key_fixed] = value
        else:
            continue
            
        if (
            lora_tfr_A_weight is not None and 
            lora_tfr_B_weight is not None and 
            lora_ref_A_weight is not None and 
            lora_ref_B_weight is not None and 
            gate_A_weight is not None and 
            gate_B_weight is not None and
            gate_offset is not None
        ):
            ref_weight = (lora_ref_B_weight @ lora_ref_A_weight) * ref_scaling
            delta_weight = (lora_tfr_B_weight @ lora_tfr_A_weight) * tfr_scaling
            # Gate g_i = sigmoid(raw gate + scheduled offset + learned offset),
            # scaled by the warmup schedule
            gate_weight = torch.sigmoid(
                (gate_B_weight @ gate_A_weight) + offset_scheduled + gate_offset
            ) * gate_alpha
            
            key_base = key_fixed.replace('.gate.', '.lora.').split('.lora.')[0] + '.weight'
            ref_state_dict[key_base] = ref_weight
            delta_state_dict[key_base] = delta_weight
            gate_state_dict[key_base] = gate_weight
            
            lora_tfr_A_weight = None
            lora_tfr_B_weight = None
            lora_ref_A_weight = None
            lora_ref_B_weight = None
            gate_A_weight = None
            gate_B_weight = None
            gate_offset = None
            
    return delta_state_dict, ref_state_dict, gate_state_dict

legamex_task_delta, legamex_task_ref, legamex_task_gate = fix_legamex_state_dict(
    legamex_task_state_dict, legamex_task_config, LEGAMEX_TASK_CKPT_STEP)
legamex_lang_delta, legamex_lang_ref, legamex_lang_gate = fix_legamex_state_dict(
    legamex_lang_state_dict, legamex_lang_config, LEGAMEX_LANG_CKPT_STEP)
```

```python
# Averaged reference-LoRA delta (refavg). Both checkpoints load the SAME frozen
# reference LoRA (REF_LORA_ID, kept frozen during training), so
# d_ref_task == d_ref_lang up to numerical noise; we still average the two
# reconstructed ref deltas so the 'averaged' name is literal and stays correct
# if a future checkpoint pair ever used a different ref config. The ref GATE
# (mean_i (1 - c_i)) is derived from the ADDITION_TYPE coefficients in the next
# cell so the ref strength stays consistent with the delta scaling.
legamex_ref_avg_delta = OrderedDict()
task_ref_keys = list(legamex_task_ref.keys())
assert task_ref_keys == list(legamex_lang_ref.keys()), \
    "Task/lang reference deltas must have the same keys (and order)."
for key in task_ref_keys:
    legamex_ref_avg_delta[key] = (legamex_task_ref[key] + legamex_lang_ref[key]) / 2.0
```

```python
def normalize_gate_state_dicts(
    gate_state_dicts: list[OrderedDict[str, torch.Tensor]]
) -> list[OrderedDict[str, torch.Tensor]]:
    """'gna': normalize each model's gates so they sum to 1 elementwise.

    w_i = g_i / SUM_j g_j for every weight element, so that SUM_i w_i = 1. This
    is the "gate-normalized" step: the models' gates (which independently lie in
    (0, 1) and may sum to < or > 1) become a convex weighting.
    """
    key_lists = [list(gate_state_dict.keys()) for gate_state_dict in gate_state_dicts]
    assert all(key_list == key_lists[0] for key_list in key_lists), \
        "All gate_state_dicts must have the same keys (and order)."
    
    EPS = 1e-12
    gate_state_dicts_normalized = []
    for key in key_lists[0]:
        values = [gate_state_dict[key] for gate_state_dict in gate_state_dicts]
        sum_ = torch.stack(values).sum(dim=0)
        n_degenerate = int((sum_ <= EPS).sum())
        if n_degenerate:
            print(
                f"Warning: {n_degenerate} weight(s) have ~zero total gate at "
                f"'{key}'; clamping to {EPS} to avoid NaN."
            )
        sum_ = sum_.clamp_min(EPS)
        for i, gate_state_dict in enumerate(gate_state_dicts):
            value_normalized = gate_state_dict[key] / sum_
            if len(gate_state_dicts_normalized) <= i:
                gate_state_dicts_normalized.append(OrderedDict())
            gate_state_dicts_normalized[i][key] = value_normalized
    return gate_state_dicts_normalized


def compute_merge_coefficients(
    addition_type: str,
    gate_state_dicts: list[OrderedDict[str, torch.Tensor]],
    addition_weights: list[float] | None = None,
    shrink_lambda: float = 1.0,
    moat_beta: float = 2.0,
    moat_sigma: float = 0.04,
) -> list[OrderedDict[str, torch.Tensor]]:
    """Per-model merge coefficient c_i, applied to each RAW delta d_i.

    'add' / 'avg': c_i = a_i * g_i  (gated deltas, scalar-weighted; a=[1,1]/[0.5,0.5]/[1,1])
    'gna':         c_i = g_i / SUM  (gate-normalized; SUM_i c_i = 1 elementwise)
    'nga':         c_i = 1.0        (no-gate addition; raw deltas at full strength)
    'shrink':      c_i = 0.5 + shrink_lambda*(g_i - 0.5)  (gates pulled shrink_lambda-fraction toward 0.5)
    'moat':        task only: c = g + beta*(g-0.5)*exp(-(g-0.5)^2/(2 sigma^2))  (carve the 0.5 spike, keep tails+curve)
    """
    if addition_type in ('add', 'avg'):
        assert addition_weights is not None and len(addition_weights) == len(gate_state_dicts), \
            f"'{addition_type}' needs one scalar weight per model."
        return [
            OrderedDict({k: w * gate for k, gate in gate_state_dict.items()})
            for w, gate_state_dict in zip(addition_weights, gate_state_dicts)
        ]
    elif addition_type == 'gna':
        return normalize_gate_state_dicts(gate_state_dicts)
    elif addition_type == 'nga':
        # 'nga' (no-gate addition): c_i = 1.0 for every weight, so each model's
        # RAW transfer delta is added at full strength
        # (W_merged = W_base + d_task + d_lang). The all-ones dicts keep the
        # same keys as the gate dicts so the merge loop's per-name coefficient
        # lookup still works (qa_outputs never appear here and are special-cased
        # in the merge loop anyway).
        return [
            OrderedDict({k: torch.ones_like(gate) for k, gate in gate_state_dict.items()})
            for gate_state_dict in gate_state_dicts
        ]
    elif addition_type == 'shrink':
        # 'shrink': pull each gate toward the fair middle (0.5) by a factor
        # shrink_lambda:  c_i = 0.5 + shrink_lambda * (g_i - 0.5).
        #   shrink_lambda=1.0 -> gates exactly as learned (same as 'add')
        #   shrink_lambda=0.5 -> tails pulled halfway back to 0.5
        #   shrink_lambda=0.0 -> every gate EXACTLY 0.5 (uniform half-strength)
        # Only the spread changes; the 0.5 mass stays identical, so this isolates
        # whether the left/right gate tails matter.
        assert 0.0 <= shrink_lambda <= 1.0, \
            f"shrink_lambda must be in [0, 1], got {shrink_lambda}."
        return [
            OrderedDict({k: 0.5 + shrink_lambda * (gate - 0.5) for k, gate in gate_state_dict.items()})
            for gate_state_dict in gate_state_dicts
        ]
    elif addition_type == 'moat':
        # 'moat': remove ONLY the 0.5 (dead-center) spike of the TASK gate, keeping
        # the left/right tails and the smooth curve between them essentially intact:
        #   c = g + moat_beta * (g - 0.5) * exp(-(g - 0.5)**2 / (2 * moat_sigma**2))
        # The exp() factor is a window: far from 0.5 it is ~0 so the map is the
        # identity (tails and bridging curve preserved); near 0.5 it pushes mass
        # OUTWARD, digging a smooth canyon in place of the artificial sigmoid(0)
        # spike (the fair-middle offset pins g=0.5 even when B_g@A_g~0).
        # moat_beta <= 2.2 keeps the map strictly increasing (c'(g) >= 1 - 0.446*beta
        # > ~0.02), preserving weight order. Applied to the TASK gate only (index 0);
        # the LANG gate is returned unchanged (its gates are all concentrated at 0.5
        # with no left/right structure to preserve).
        assert len(gate_state_dicts) == 2, "'moat' expects exactly two gate state dicts ([task, lang])."
        assert 0.0 < moat_sigma, f"moat_sigma must be > 0, got {moat_sigma}."
        assert 0.0 <= moat_beta <= 2.2, \
            f"moat_beta must be in [0, 2.2] so the map stays monotone, got {moat_beta}."
        task_coeffs = OrderedDict()
        for key, gate in gate_state_dicts[0].items():
            deviation = gate - 0.5
            carve = moat_beta * deviation * torch.exp(
                -(deviation ** 2) / (2.0 * moat_sigma ** 2)
            )
            task_coeffs[key] = (gate + carve).clamp(0.0, 1.0)
        return [task_coeffs, OrderedDict(gate_state_dicts[1])]
    else:
        raise ValueError(f"Unknown addition type: {addition_type}. Use 'add', 'avg', 'gna', 'nga', 'shrink', or 'moat'.")


# ORDERING: keep this list in the SAME order as fixed_state_dicts in the merge
# cell below ([task, lang, ref-avg]), so zip() pairs each delta with its own
# coefficient.
merge_coefficient_state_dicts = compute_merge_coefficients(
    ADDITION_TYPE,
    [legamex_task_gate, legamex_lang_gate],
    addition_weights,
    SHRINK_LAMBDA,  # used by 'shrink'
    MOAT_BETA,  # used by 'moat'
    MOAT_SIGMA,  # used by 'moat'
)
merge_coefficient_state_dicts = [
    {k.removeprefix('base_model.'): v for k, v in state_dict.items()}
    for state_dict in merge_coefficient_state_dicts
]

# Averaged gated reference component (refavg): merge the shared reference-LoRA
# delta ONCE, gated by the average of the two models' ref gates (1 - c_i),
# with c_i each model's ADDITION_TYPE coefficient computed above:
#   W_merged = W_base + SUM_i c_i*d_tfr_i + (mean_i (1 - c_i)) * d_ref_avg
# This ties the ref strength to the chosen merge type: 'add' -> 1-avg(g),
# 'avg' -> 1-0.5*avg(g), 'gna' -> exactly 0.5 (the c_i are convex),
# 'nga' -> 0 (ref omitted, same as the full notebook), 'shrink'/'moat' -> the
# transformed gates averaged.
ref_avg_gate = OrderedDict()
coeff_key_lists = [
    list(state_dict.keys()) for state_dict in merge_coefficient_state_dicts
]
assert all(key_list == coeff_key_lists[0] for key_list in coeff_key_lists), \
    "All merge coefficient dicts must have the same keys (and order)."
for key in coeff_key_lists[0]:
    ref_avg_gate[key] = torch.stack([
        1.0 - coeff_state_dict[key]
        for coeff_state_dict in merge_coefficient_state_dicts
    ]).mean(dim=0)
merge_coefficient_state_dicts.append(ref_avg_gate)
```

```python
# Load the base model
base_model = AutoModelForQuestionAnswering.from_pretrained(MODEL_ID)
```

# Merge with Weighted Addition / Averaging


```python
# Weighted merge over the models' RAW LegameX deltas PLUS the averaged gated
# reference component (refavg).
#
#   W_merged = W_base + SUM_i [ c_i * d_tfr_i ] + (mean_i (1 - c_i)) * d_ref_avg
#
# where d_tfr_i is each model's RAW transfer delta (from fix_legamex_state_dict),
# c_i its per-model merge coefficient (from compute_merge_coefficients):
#   'add': c_i = g_i        sum of gated deltas
#   'avg': c_i = a_i * g_i  scalar-weighted average of gated deltas
#   'gna': c_i = g_i / SUMg Gate-Normalized Weighted Averaging (convex combination)
#   'nga': c_i = 1.0         no-gate addition (raw deltas at full strength)
#   'shrink': c_i = 0.5 + SHRINK_LAMBDA*(g_i - 0.5)  gates pulled SHRINK_LAMBDA-fraction toward 0.5
#   'moat': task gates only: c = g + beta*(g-0.5)*exp(-(g-0.5)^2/(2 sigma^2))  carve the 0.5 spike, keep tails+curve
#
# ...and d_ref_avg is the MEAN of the two models' raw reference-LoRA deltas
# (both loaded frozen from the shared reference LoRA, identical up to numerical
# noise; see the averaged-delta cell). The transfer half c_i*d_tfr_i mirrors the
# gated blend W_eff = W_base + g*d_tfr + (1-g)*d_ref from LegamexLora.forward /
# merge_and_unload (legamex_training_xlmr.ipynb); the ref delta itself is merged
# ONCE (not once per model - it is the same frozen shared branch in both
# checkpoints) at the average of the two models' ref gates (1 - c_i). This is
# the 'averaged gated ref component' that this notebook adds to the plain merge.
#
# This mirrors the sequential PEFT merge in notebooks/lora_addition_xlmr.ipynb
# (W <- W + w * (B @ (w-scaled) A)), but with the deltas precomputed and the
# per-model coefficient derived from the learned gate.
#
# Ordering: fixed_state_dicts and merge_coefficient_state_dicts are both
# [task, lang, ref-avg], so zip() pairs each delta with its own coefficient.
fixed_state_dicts = [
    legamex_task_delta,
    legamex_lang_delta,
    legamex_ref_avg_delta,
]

# Defensive normalization: strip a 'base_model.' prefix if present
# (current checkpoints do not use one; older versions did).
fixed_state_dicts = [
    {k.removeprefix('base_model.'): v for k, v in state_dict.items()}
    for state_dict in fixed_state_dicts
]

current_model = base_model
for model_idx, (delta_state_dict, coeff_state_dict) in enumerate(zip(fixed_state_dicts, merge_coefficient_state_dicts)):
    n_deltas, n_heads = 0, 0
    with torch.no_grad():
        for name, param in current_model.named_parameters():
            if name not in delta_state_dict:
                continue
            value = delta_state_dict[name].to(device=param.device, dtype=param.dtype)
            if 'qa_outputs' in name:
                # qa_outputs entries are FULL trained weights, not deltas:
                # the base XLM-R QA head is randomly initialized, so we copy
                # the trained head instead of adding it on top of noise.
                # (Only the squad/task LegameX provides a QA head; the
                #  wikipedia lang LegameX is FEATURE_EXTRACTION and has none.)
                param.data.copy_(value)
                n_heads += 1
            else:
                coeff = coeff_state_dict[name].to(device=param.device, dtype=param.dtype)
                param.data += coeff * value
                n_deltas += 1
    if model_idx == len(fixed_state_dicts) - 1:
        print(f"[refavg] reference: added {n_deltas} ref deltas (coefficient mean_i(1 - c_i))")
    else:
        print(f"[{ADDITION_TYPE}] model: added {n_deltas} deltas (coefficient c_i), copied {n_heads} qa_outputs tensors")

merged_model = current_model
```

```python
# Sanity check
for i in range(11):
    print(f"Merged model layer-{i} attention value weight norm:", merged_model.roberta.encoder.layer[i].attention.self.value.weight.norm().item())
print()
print("Merged model qa_outputs weight norm:", merged_model.qa_outputs.weight.norm().item())
print("Merged model qa_outputs bias norm:", merged_model.qa_outputs.bias.norm().item())

# 'nga' invariant: every merge coefficient must be EXACTLY 1.0 (learned gates omitted)
if ADDITION_TYPE == 'nga':
    first_key = next(iter(merge_coefficient_state_dicts[0]))
    c0 = merge_coefficient_state_dicts[0][first_key]
    c1 = merge_coefficient_state_dicts[1][first_key]
    print()
    print(f"NGA coefficients at '{first_key}': task min={c0.min().item():.6f} "
          f"max={c0.max().item():.6f} | lang min={c1.min().item():.6f} "
          f"max={c1.max().item():.6f}  (expect 1.0)")

# 'shrink' report: lambda used and the resulting coefficient range
if ADDITION_TYPE == 'shrink':
    first_key = next(iter(merge_coefficient_state_dicts[0]))
    c0 = merge_coefficient_state_dicts[0][first_key]
    c1 = merge_coefficient_state_dicts[1][first_key]
    print()
    print(f"SHRINK lambda={SHRINK_LAMBDA}: coeff at '{first_key}' "
          f"task min={c0.min().item():.6f}, max={c0.max().item():.6f} | lang "
          f"min={c1.min().item():.6f}, max={c1.max().item():.6f} "
          f"(lambda=0 -> all 0.5; lambda=1 -> as learned)")

# 'moat' report: how much of the 0.5 spike got carved on the TASK side (lang untouched)
if ADDITION_TYPE == 'moat':
    first_key = next(iter(merge_coefficient_state_dicts[0]))
    c0 = merge_coefficient_state_dicts[0][first_key]  # task: moated
    c1 = merge_coefficient_state_dicts[1][first_key]  # lang: untouched
    in_mid = lambda t: float(((t >= 0.48) & (t <= 0.52)).float().mean())
    print()
    print(f"MOAT beta={MOAT_BETA:g} sigma={MOAT_SIGMA:g}: task coeff at '{first_key}' "
          f"min={c0.min().item():.4f}, max={c0.max().item():.4f}, "
          f"% in [0.48,0.52] = {in_mid(c0)*100:.1f}  | lang (untouched): "
          f"% in [0.48,0.52] = {in_mid(c1)*100:.1f}  (expect task % well below lang %)")

# 'gna' invariant: the merge coefficients must sum to 1 at every weight element,
# i.e. the merged delta is a true convex combination of the two models' deltas.
# ('add'/'avg' coefficients sum to SUM g_i / 0.5*SUM g_i, which is not 1 in general.)
if ADDITION_TYPE == 'gna':
    first_key = next(iter(merge_coefficient_state_dicts[0]))
    coeff_sum = (merge_coefficient_state_dicts[0][first_key]
                 + merge_coefficient_state_dicts[1][first_key])
    print()
    print(f"GNA coefficient sum (task+lang) at '{first_key}': "
          f"min={coeff_sum.min().item():.6f}, max={coeff_sum.max().item():.6f}  (expect 1.0)")
```
